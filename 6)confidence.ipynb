## Confidence Interval

```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
sample = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29 ]
```

```python
n = len(sample)
```

```python
sam_mean =round( np.mean(sample),3)  
sam_mean
```

```python
sam_std=round(np.std(sample),3)
sam_std
```

```python
std_error = sam_std/np.sqrt(n)
```

```python
confi = 0.99
alpha= 1- confi
```

```python
t_cri = stats.t.ppf(1-alpha/2,df=n-1)
```

```python
margin_error = t_cri* std_error
```

```python
upper_limit= sam_mean+margin_error
lower_limit=sam_mean-margin_error
```

```python
print(f"the sample mean  is :",sam_mean)
print(f"the sample standard deviation  is :",sam_std)
print(f"the stanndard errror  is :",std_error)
print(f"the upper limit is :",upper_limit)
print(f"the lower  limit is :",lower_limit)

```

```python
## first we find the mean of the data
## later w efind the std
## then we find the standard error
## we find the critical using the inbuilt stats
## then we find the margin error
```

## Hypothesis testing

```python
# 1  null : There is no increase in the average weekly operating cost.
#    alternate: The average weekly operating cost has increased.
```

```python
import numpy as np
from scipy.stats import norm
```

```python
pop_mean= 4000
sam_mean=3050
pop_std= 125
n=25
z_score = (sam_mean - pop_mean)/(pop_std/np.sqrt(n))
z_score
```

```python
p_val = 1 - norm.cdf(z_score)
print(p_val)
print(z_score)
alpha= 0.05
```

```python
if p_val < alpha:
    print("reject null hypothesis")
else:
    print("fail to reject null hypothesis")    
```
