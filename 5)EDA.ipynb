##### The strongest connection was with ASTV (abnormal short-term variability), which had a correlation of 0.42 with NSP. This means that when short-term variability is abnormal, the fetus is more likely to be in a distressed state. ALTV (abnormal long-term variability) also showed a strong relationship with NSP at 0.37. These findings make sense because doctors often monitor heart rate variability to check if the baby is doing well.m

####  Data Quality: The dataset had some missing values, which I filled using average values. I also removed duplicate rows to keep the data clean.

#### Most Important Variables: The analysis showed that ASTV (abnormal short-term variability) and ALTV (abnormal long-term variability) have the strongest connection with fetal health. Higher values of these variables are linked to abnormal fetal states.

#### Accelerations are a Good Sign: AC (accelerations) showed a negative relationship with NSP. This means when there are more accelerations, the baby is usually healthy. Fewer accelerations could be a warning sign.

#### Decelerations are Warning Signs: DP (prolonged decelerations) showed a positive link with NSP. More decelerations mean more risk of fetal distress.

#### Some Variables May Not Help: Variables like FM (fetal movements) and DS (short decelerations) showed very weak correlations with NSP. These might not be very useful for predicting fetal health.

#### Variability Measures are Connected: ASTV, ALTV, MSTV, and Width are somewhat related to each other. This means they measure similar things, so we might not need all of them in a prediction model.
