# 2 Dimensionality Reduction with PCA: 

```python
df['Type'].value_counts()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
x=df.drop('Type',axis =1)
y=df['Type']
```

```python
scaler = StandardScaler()
x_scal = scaler.fit_transform(x)
```

```python
from sklearn.decomposition import PCA

pca = PCA()
X_pca = pca.fit_transform(x_scal)
```

```python
import numpy as np

print(np.cumsum(pca.explained_variance_ratio_))
```

```python
print(pca.explained_variance_ratio_)
```

#### so if i keep ncomponenet then we get variance upto 83

```python
pca = PCA(n_components=6)
X_pca = pca.fit_transform(x_scal)
```

```python
X_pca.head()
```

```python
import pandas as pd

X_pca = pd.DataFrame(
    X_pca,
    columns=['PC1','PC2','PC3','PC4','PC5','PC6']
)
```

# clustering on original data

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
inertia = []
k_range = range(1,11)
```

```python
for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(x_scal)
    inertia.append(kmeans.inertia_)
```

```python
plt.plot(k_range, inertia, marker='o')
```

```python
kmeans_final = KMeans(n_clusters=3, random_state=42)
cluster_labels = kmeans_final.fit_predict(x_scal)
df['cluster'] = cluster_labels
```

```python
from sklearn.metrics import silhouette_score

score = silhouette_score(x_scal, cluster_labels)
print(score)
```

```python
df.head()
```

```python
sns.scatterplot(
    x=df['Malic'],
    y=df['Alcohol'],
    hue=df['cluster'],
    palette='viridis'
)
plt.show()
```

```python
sns.scatterplot(
    x=df['Ash'],
    y=df['Alcohol'],
    hue=df['cluster'],
    palette='viridis'
)
plt.show()
```

# clustering for pca dataset

```python
for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_pca)
    inertia.append(kmeans.inertia_)
```

```python
print(X_pca.shape)
```

```python
kmeans_final = KMeans(n_clusters=3, random_state=42)
cluster_labels = kmeans_final.fit_predict(X_pca)
X_pca['cluster'] = cluster_labels
```

```python
from sklearn.metrics import silhouette_score

score = silhouette_score(X_pca, cluster_labels)
print(score)
```

```python
sns.scatterplot(
    x=df['Malic'],
    y=df['Alcohol'],
    hue=X_pca['cluster'],
    palette='viridis'
)
plt.show()
```

```python
sns.scatterplot(
    x=df['Ash'],
    y=df['Alcohol'],
    hue=X_pca['cluster'],
    palette='viridis'
)
plt.show()
```

#### the  silhouette score  for the pca dataset is 38 % but for original it is just 28% so PCA helps in Clustering

#### Both datasets produced meaningful clusters; however, the PCA-transformed data resulted in better cluster separation

#### Dimensionality reduction using PCA positively impacted clustering performance. By reducing the number of features while retaining most of the important information, PCA reduced noise and redundancy in the data

#### PCA helps simplify high-dimensional datasets by reducing redundant information while preserving the most important patterns. Clustering techniques such as K-Means  help identify groups of similar observations, making them useful for customer segmentation, marketing strategies, anomaly detection, and decision-making in business applications.
